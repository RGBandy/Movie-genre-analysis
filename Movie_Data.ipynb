```python
df = df.drop(columns=['adult','backdrop_path', 'original_language', 'poster_path','softcore', 'overview', 'video','id'], axis=1)
df.head(10)
```

```python
genre_url = 'https://api.themoviedb.org/3/genre/movie/list'
genre_responce = requests.get(genre_url, headers=headers)
genre_data = genre_responce.json()
genre_data
```

```python
genre_dict = {genre['id']: genre['name']
              for genre in genre_data['genres']
             }
genre_dict
```

```python
def get_genre_names(genre_ids):
    return [genre_dict[genre_id] for genre_id in genre_ids]

df['genres'] = df['genre_ids'].apply(get_genre_names)
df
```

```python
df['release_date'] = pd.to_datetime(df['release_date'])
df['year'] = df['release_date'].dt.year
```

```python
#used method explode so I can group generes later

genre_df = df.explode("genres")

genre_df.head()
```

```python
genre_df = genre_df.rename(columns={'genres':'genre'})
genre_df = genre_df[['genre_ids','title','year','genre','vote_average','vote_count','popularity']]
genre_df = genre_df.reset_index(drop=True)
genre_df
```

```python
genre_df.to_csv('tbmd_movies.csv',index = False)
```
